```python
MAROON = "#B51A24"
DARK_MAROON = "#8B1520"
ORANGE = "#E56B05"
BRIGHT_ORANGE = "#F28C00"
YELLOW = "#F9D212"
LIGHT_YELLOW = "#F9E98A"
GOLD = "#DDAA0B"
GREEN = "#3D9B14"
DARK_GREEN = "#318418"
LEAF_GREEN = "#2B7614"
VEIN_GREEN = "#739411"
LIGHT_GREEN = "#8CC52F"
WHITE = "#F6F4E8"
CREAM = "#F5E7B0"
BROWN = "#6A3506"
DARK_BROWN = "#3F240D"
RED = "#C64205"
PINK = "#E865A4"
PURPLE = "#9C1F63"
BLACK = "#1F260F"
BG_COLOR = "#EFEDE7"
background = rectangle(w=300, h=300, fill=BG_COLOR, stroke="none")
show(background)
```

```python
leaf1 = ellipse(x=-70,w=160,h=120,stroke="none",fill=DARK_GREEN)
leaf2 = rectangle(x=20,w=200,h=120,stroke="none",fill=DARK_GREEN)
leaf3 = line(x1=-150,y1=0,x2=110,y2=0,stroke=VEIN_GREEN,stroke_width=4)
upper_veins = line(x1=105,y1=-5,x2=105,y2=-55,stroke=VEIN_GREEN,stroke_width=2) | repeat(19,translate(x=-10,y=0))
lower_veins = line(x1=105,y1=5,x2=105,y2=55,stroke=VEIN_GREEN,stroke_width=2) | repeat(21,translate(x=-10,y=0))
left_upper = line(x1=-85,y1=0,x2=-85,y2=-48,stroke=VEIN_GREEN,stroke_width=2) | repeat(6,translate(x=-20,y=0) | scale(0.92))
left_lower = line(x1=-85,y1=0,x2=-85,y2=48,stroke=VEIN_GREEN,stroke_width=2) | repeat(6,translate(x=-20,y=0) | scale(0.92))
leaf = (leaf1 +leaf2 +leaf3 +upper_veins +lower_veins +left_upper +left_lower) | translate(x=10,y=0)
show(background,leaf)
```

```python
def salt(n=15, r=1, a=25):

    c = ellipse(w=5,h=3,fill=WHITE,stroke="none")
    temp = c
    for i in range(n):
        r1 = (c| translate(random(-8, 8), random(-5, 5))| rotate(random(360)))
        temp = temp + r1
    return temp | repeat(r, rotate(a))  | translate(-90, 45)
salt1 = salt(15, 1, 25)
show(background,leaf,salt1)
```

```python
def banana_chips(n=8, r=1, a=45):
    c = (ellipse(w=12,h=9,fill="#F2B705",stroke="none")+ circle(r=2.2,fill="#D88900",stroke="none"))
    temp = c
    for i in range(n):
        r1 = (c| translate(random(-10, 10), random(-8, 8))| rotate(random(360)))
        temp = temp + r1
    return temp | repeat(r, rotate(a)) | translate(-70, 45)
banana = banana_chips(5, 1, 45)
show(background,leaf,salt1,banana)
```

```python
def sharkaravaratti(n=7, r=1, a=35):
    c = (ellipse(w=18,h=9,fill="#9A5A18",stroke="none")+ ellipse(w=14,h=6,fill="#B87322",stroke="none"))
    temp = c
    for i in range(n):
        r1 = (c| translate(random(-10, 10), random(-6, 6))| rotate(random(-35, 35)))
        temp = temp + r1
    return temp | repeat(r, rotate(a)) | translate(-45, 45)
sharkara = sharkaravaratti(5, 1, 35)
show(background,leaf,salt1,banana,sharkara)
```

```python
def pickle(n=8, r=1, a=30):
    c = (ellipse(w=10,h=7,fill="#C83B20",stroke="none")+ circle(r=2,fill="#8F2418",stroke="none" ))
    temp = c
    for i in range(n):
        r1 = (c| translate(random(-9, 9), random(-6, 6))| rotate(random(360)))
        temp = temp + r1
    return temp | repeat(r, rotate(a)) | translate(-25, 45)
pickle1 = pickle(6, 1, 30)
show(background,leaf,salt1,banana,sharkara,pickle1)
```

```python
def pappadam():
    p1 = point(x=-28, y=-10)
    p2 = point(x=-22, y=-22)
    p3 = point(x=-8, y=-28)
    p4 = point(x=10, y=-26)
    p5 = point(x=25, y=-18)
    p6 = point(x=30, y=-5)
    p7 = point(x=26, y=12)
    p8 = point(x=15, y=23)
    p9 = point(x=-3, y=27)
    p10 = point(x=-20, y=20)
    p11 = point(x=-29, y=8)
    base = polygon([p1,p2,p3,p4,p5,p6,p7,p8,p9,p10,p11],fill="#F4D96B",stroke="none")
    bubble1 = ellipse(x=-12, y=-10,w=18, h=13,fill="#FFF0A8",stroke="none")
    bubble2 = ellipse(x=7, y=-13,w=20, h=12,fill="#FFE98A",stroke="none")
    bubble3 = ellipse(x=10, y=6,w=19, h=14,fill="#FFF0A8",stroke="none")
    bubble4 = ellipse(x=-13, y=10,w=17, h=12,fill="#FFE98A",stroke="none")
    bubble5 = ellipse(x=-2, y=0,w=14, h=10,fill="#FCE88D",stroke="none")
    return base + bubble1 + bubble2 + bubble3 + bubble4 + bubble5  | translate(-95,-20)
pappadam1 = pappadam()
show(background,leaf,salt1,banana,sharkara,pickle1,pappadam1)
```

```python
def kondattam_piece():
    p1 = point(x=-8, y=-3)
    p2 = point(x=-5, y=-6)
    p3 = point(x=5, y=-4)
    p4 = point(x=9, y=0)
    p5 = point(x=5, y=4)
    p6 = point(x=-4, y=5)
    p7 = point(x=-8, y=2)
    return polygon([p1, p2, p3, p4, p5, p6, p7],fill="#8F2118",stroke="none")
def kondattam(n=5):
    c = kondattam_piece()
    temp = c
    for i in range(n):
        r1 = (c| translate(random(-10, 10), random(-8, 8))| rotate(random(360)))
        temp = temp + r1
    return temp  | translate(-95, 23)
kondattam1 = kondattam(5)
show(background,leaf,salt1,banana,sharkara,pickle1,pappadam1,kondattam1)
```

```python
def kaalan(n=7):
    base = ellipse(w=24,h=18,fill="#F3E6A2",stroke="none")
    temp = base
    for i in range(n):
        banana = ellipse(w=7,h=5,
  fill="#D6B83F",stroke="none")| translate(random(-8, 8),random(-5, 5))| rotate(random(360))
        temp = temp + banana
    for i in range(4):
        chilli = (ellipse(w=5,h=2.5,fill="#71852B",stroke="none")| translate(random(-8, 8),random(-5, 5))| rotate(random(360)))
        temp = temp + chilli
    return temp  | translate(5, 40)
kaalan1  =kaalan(7)
show(background,leaf,salt1,banana,sharkara,pickle1,pappadam1,kondattam1,kaalan1)
```

```python
def olan(n=8):
    base = ellipse(w=26,h=18,fill="#F5E9B5",stroke="none")
    temp = base
    for i in range(n):
        veg = (ellipse(w=7,h=3.5,fill="#7A9A35",stroke="none")| translate(random(-9, 9),random(-5, 5))| rotate(random(360)))
        temp = temp + veg
    for i in range(5):
        coconut = (circle(r=2,fill="#FFF5CF",stroke="none")| translate(random(-9, 9),random(-5, 5)))
        temp = temp + coconut
    return temp | translate(30, 40)
olan1 = olan(8)
show(background,leaf,salt1,banana,sharkara,pickle1,pappadam1,kondattam1,kaalan1,olan1)
```

```python
def thoran(n=12):
    base = ellipse(w=30,h=20,fill="#223e01",stroke="none")
    temp = base
    for i in range(n):
        green = (rectangle(w=10,h=3,fill="#3F6F22",stroke="none")| translate(random(-10, 10),random(-6, 6))| rotate(random(360)))
        temp = temp + green
    for i in range(6):
        light_green = (rectangle(w=7,h=3,fill="#9EAD3B",stroke="none")| translate(random(-9, 9),random(-6, 6))| rotate(random(360)))
        temp = temp + light_green
    for i in range(7):
        coconut = (circle(r=2,fill="#F2E5A5",stroke="none")| translate(random(-9, 9),random(-6, 6)))
        temp = temp + coconut
    return temp  | translate(110, 45)
thoran1 = thoran(12)
show(background,leaf,salt1,banana,sharkara,pickle1,pappadam1,kondattam1,kaalan1,olan1,avial,thoran1)
```

```python
def erissery(n=8):
    base = ellipse(w=30, h=20, fill="#D69B22", stroke="none")
    temp = base
    for i in range(n):
        pumpkin = (rectangle(w=8, h=5, fill="#F2C43D", stroke="none")
                   | translate(random(-9, 9), random(-6, 6))
                   | rotate(random(360)))
        temp = temp + pumpkin
    for i in range(5):
        green = (rectangle(w=6, h=3, fill="#607D25", stroke="none")
                 | translate(random(-9, 9), random(-6, 6))
                 | rotate(random(360)))
        temp = temp + green
    for i in range(6):
        coconut = (circle(r=2, fill="#F4E7A5", stroke="none")
                   | translate(random(-9, 9), random(-6, 6)))
        temp = temp + coconut
    return temp | translate(-60, 25)   # ← now outside all three loops

erissery1 = erissery(8)
show(background,leaf,salt1,banana,sharkara,pickle1,pappadam1,kondattam1,kaalan1,olan1,avial,thoran1,erissery1 )
```

```python
def sambar(n=10):
    base = ellipse(w=32,h=21,fill="#C96A22",stroke="none")
    temp = base
    for i in range(n):
        veg = (rectangle(w=7,h=5,fill="#E5A82E",stroke="none")| translate(random(-10, 10),random(-6, 6))| rotate(random(360)))
        temp = temp + veg
    for i in range(5):
        green = (rectangle(w=7,h=3,fill="#657A28",stroke="none")| translate(random(-9, 9),random(-6, 6))| rotate(random(360)))
        temp = temp + green
    for i in range(4):
        spice = (circle(r=2,fill="#8F3B1C",stroke="none")| translate(random(-10, 10),random(-6, 6)))
        temp = temp + spice
    return temp | translate(50, -15) | scale(1.2)
sambar1 = sambar(10)
show(background,leaf,salt1,banana,pickle1,pappadam1,kondattam1,kaalan1,olan1,avial,thoran1,erissery1,rice,sambar1)
```

```python
def payasam(n=10):
    bowl = circle(r=17, fill="#D8CDB0", stroke=BROWN, stroke_width=2.5)
    temp = bowl
    for i in range(n):
        strand = (rectangle(w=7, h=1.6, fill="#F3ECDA", stroke="none")
                  | translate(random(-11, 11), random(-11, 11))
                  | rotate(random(360)))
        temp = temp + strand
    for i in range(4):
        cashew = (ellipse(w=3, h=2, fill="#C98A3E", stroke="none")
                  | translate(random(-10, 10), random(-10, 10))
                  | rotate(random(360)))
        temp = temp + cashew
    for i in range(3):
        raisin = (circle(r=1.2, fill="#5C3A21", stroke="none")
                  | translate(random(-10, 10), random(-10, 10)))
        temp = temp + raisin
    for i in range(3):
        sheen = (circle(r=1, fill="#FFF6DE", stroke="none")
                 | translate(random(-9, 9), random(-9, 9)))
        temp = temp + sheen
    return temp | translate(95, -1)
payasam1 = payasam(10)
show(background,leaf,salt1,banana,pickle1,pappadam1,kondattam1,kaalan1,olan1,avial,thoran1,erissery1,rice,sambar1,payasam1)
```

```python
def ada_pradhaman(n=3):
    bowl = circle(r=17, fill="#6B2A1A", stroke="#B7BEC4", stroke_width=1.5)
    temp = bowl
    for i in range(n):
        ada = (ellipse(w=9, h=7, fill="#A9673A", stroke="#7A4526", stroke_width=0.4)
               | translate(random(-9, 9), random(-8, 8))
               | rotate(random(360)))
        temp = temp + ada
    for i in range(5):
        coconut = (rectangle(w=5, h=1.8, fill="#D9A62E", stroke="none")
                   | translate(random(-11, 11), random(-11, 11))
                   | rotate(random(360)))
        temp = temp + coconut
    for i in range(3):
        cashew = (ellipse(w=3, h=2, fill="#E8D4A0", stroke="none")
                  | translate(random(-10, 10), random(-10, 10))
                  | rotate(random(360)))
        temp = temp + cashew
    for i in range(3):
        sheen = (circle(r=1, fill="#C98A4E", stroke="none")
                 | translate(random(-9, 9), random(-9, 9)))
        temp = temp + sheen
    return temp | translate(95, -40)   # same payasam slot — adjust if you're placing it beside payasam1
ada_pradhaman1 = ada_pradhaman(3)
show(background,leaf,salt1,banana,pickle1,pappadam1,kondattam1,kaalan1,olan1,avial,thoran1,erissery1,rice,sambar1,payasam1,ada_pradhaman1)
```
